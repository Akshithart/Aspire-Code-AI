CONFIGURATION SECTIONS

```python
MODEL_NAME="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"
TEMPERATURE=1.5
MAX_TOKENS=200
```

LOAD MODEL

```python
import warnings
warnings.filterwarnings("ignore")

!pip install --upgrade sympy
!pip install --upgrade transformers

from transformers import pipeline

generator = pipeline(
    "text-generation",
    model=MODEL_NAME,
    device_map="auto"
)

print("Model Loaded Successfully")
```

GENERATION FUNCTION

```python
def generate_response(prompt):
    # Cricket-related keywords
    cricket_keywords = [
        "cricket", "batting", "bowling", "fielding", "wicket",
        "batsman", "bowler", "spin", "googly", "yorker", "cover drive",
        "leg spin", "off spin", "umpire", "run out", "stump"
    ]

    # Check if prompt is cricket-related
    if not any(word.lower() in prompt.lower() for word in cricket_keywords):
        return "⚠️ I am a Cricket Coach. I only answer cricket-related questions."

    messages = [
        {"role": "system", "content": "You are a professional cricket coach. Only answer cricket-related queries."},
        {"role": "user", "content": prompt}
    ]

    try:
        text = generator.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )
    except Exception:
        text = prompt

    output = generator(
        text,
        max_new_tokens=MAX_TOKENS,
        temperature=TEMPERATURE,
        do_sample=True,
        return_full_text=False
    )

    return output[0]["generated_text"]
```

TEST PROMPT


```python
reply = generate_response("Suggest batting drills for beginners")
print("Cricket Coach:", reply)
```

GRADIO INTERFACE

```python
import gradio as gr

with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("# 🏏 Cricket Coach - Powered by DeepSeek")

    prompt = gr.Textbox(
        label="Ask your Cricket Coach",
        lines=6,
        placeholder="e.g., How to improve against spin bowling?"
    )

    btn = gr.Button("🚀 Get Coaching Advice")

    output = gr.Textbox(
        label="Coach Response",
        lines=12
    )

    btn.click(
        fn=generate_response,  # your DeepSeek function
        inputs=prompt,
        outputs=output
    )

demo.launch(share=True)
```
